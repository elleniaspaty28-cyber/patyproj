```python
import pandas as pd
import matplotlib.pyplot as plt

# Data for Chart 1
data_claiming = {
    'Age': ['62', '63', '64', '65', '66 (FRA)', '67', '68', '69', '70'],
    'Percent': [41, 8, 9, 11, 21, 4, 2, 1, 3]
}
df1 = pd.DataFrame(data_claiming)

# Plot
plt.figure(figsize=(8, 4))
plt.bar(df1['Age'], df1['Percent'], color='#6baed6', edgecolor='black', width=0.5)
plt.xlabel('Self-Reported Claiming Age')
plt.ylabel('Percent of Claimants (%)')
plt.ylim(0, 50)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add data numbers above each bar
for i, val in enumerate(df1['Percent']):
    plt.text(i, val + 1, f"{val}%", ha='center', fontsize=9)

plt.tight_layout()
plt.savefig('claiming_dist.png', dpi=300)
plt.show()
```

```python
# Data for Chart 2
data_wealth = {
    'Quartile': ['Q1 (Lowest)', 'Q2', 'Q3', 'Q4 (Highest)'],
    'Percent': [58, 44, 32, 21]
}
df2 = pd.DataFrame(data_wealth)

# Plot
plt.figure(figsize=(7, 4))
plt.bar(df2['Quartile'], df2['Percent'], color='#6baed6', edgecolor='black', width=0.5)
plt.xlabel('Household Liquid Wealth Quartile')
plt.ylabel('Percent Claiming Early at Age 62 (%)')
plt.ylim(0, 70)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add data numbers above each bar
for i, val in enumerate(df2['Percent']):
    plt.text(i, val + 1, f"{val}%", ha='center', fontsize=9)

plt.tight_layout()
plt.savefig('early_claim_wealth.png', dpi=300)
plt.show()
```
